# Introduction

Over the past two decades, board games have moved from a niche hobby to a rapidly growing entertainment market, with global sales estimated in the tens of billions of dollars and steady year-over-year growth (Jolin; Peiser). This resurgence has been driven in part by modern “Eurogames” and strategy titles, as well as by online communities where players rate, review, and discuss games. One of the most influential of these communities is **BoardGameGeek (BGG)**, a large, public database where hobbyists log plays, assign ratings, and tag games by mechanics, themes, and complexity. Because BGG users tend to be highly engaged players rather than casual consumers, its ratings offer a valuable window into what this audience considers a “good” game.

**Our project uses a comprehensive dataset of over 17,000 board games compiled from BGG and Amazon by a Kaggle contributor**. The dataset includes publication year, recommended age and player counts, playtime, a 1–5 “complexity” (weight) score, detailed rating distributions, and various popularity metrics such as number of ratings, ownership counts, and total plays. After cleaning and restricting to modern games (released from 1990 onward) and dropping entries with missing key variables, we analyze a **final sample of 1,975 games** with complete information.

Theoretically, prior work on games and motivation suggests that **challenge is a core driver of enjoyment.** Malone’s classic framework on intrinsically motivating instruction, developed partly from studies of early computer games, identifies challenge as one of the main factors that make an activity engaging (Malone). In tabletop games, BGG’s “complexity” score can be viewed as a rough measure of this challenge: capturing how many rules there are, how many decisions players must make, and how strategically demanding the game is. At the same time, average playtime represents the time cost of a single session. A game can be deep and satisfying, but if it takes too long to play, some players might rate it less favorably compared with shorter alternatives that offer similar strategic depth.

We suspect complexity (weight) is linearly correlated with average rating, so we will perform a linear regression to confirm.

```python
var_x = 'complexity_std'
var_y = 'avg_rating_std'

df_reg = df[[var_x, var_y]].dropna()

X = df_reg[[var_x]].values
y = df_reg[var_y].values

reg = LinearRegression()
reg.fit(X, y)
y_pred = reg.predict(X)

slope = reg.coef_[0]
intercept= reg.intercept_
r_squared= r2_score(y, y_pred)
correlation= df_reg[var_x].corr(df_reg[var_y])

print(f"Selected Variables: {var_x} (Independent) vs {var_y} (Dependent)")
print(f"Correlation Coefficient: {correlation:.4f}")
print(f"R-squared: {r_squared:.4f}")
print(f"Regression Equation: y = {slope:.4f}x + {intercept:.4f}")

plt.figure(figsize=(10, 6))
plt.scatter(df_reg[var_x], df_reg[var_y], alpha=0.2, s=10, color='teal', label='Data Points')
plt.plot(df_reg[var_x], y_pred, color='red', linewidth=2, label=f'Fit: y={slope:.2f}x+{intercept:.2f}')

plt.title(f'Linear Regression: {var_x.capitalize()} vs {var_y.replace("_", " ").title()}', fontsize=14)
plt.xlabel(var_x.capitalize(), fontsize=12)
plt.ylabel(var_y.replace("_", " ").title(), fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig('linear_regression_plot.png')
plt.show()
```

#Preregistration Statement 1
##Hypothesis 1:
***Games with higher complexity (on the weight scale from 1 to 5) tend to have higher average ratings, even after accounting for the number of ratings they receive.***

Game research has long argued that challenge is a core ingredient of enjoyment. Malone’s classic work on intrinsically motivating instruction, based partly on early computer games, identifies “challenge” as one of the main factors that make activities engaging (Malone). In modern tabletop games, complexity (or “weight”) often reflects how challenging and strategically demanding a game is. Since BoardGameGeek’s user base skews toward hobbyists who seek depth and strategy, we expect that more complex games will tend to receive higher average ratings than simpler ones.


**Null Hypothesis (H₀):**
 Game complexity has no effect on average BGG rating after controlling for popularity (number of ratings). In a linear regression, the coefficient for complexity = 0.


**Alternative Hypothesis (H₁):**
 Game complexity has a positive effect on average BGG rating after controlling for popularity. In a linear regression, the coefficient for complexity > 0.



##Proposed Analysis:
We will run a **multiple linear regression** using *avg_rating* as the dependent variable and both *weight (complexity score)* and *log_num_ratings (log-transformed number of user ratings)* as independent variables. Before running the regression, we will also compute **Pearson and Spearman correlations** between *avg_rating* and *weight* and visualize them with a scatterplot to check for linear versus monotonic relationships.

**We will reject H₀ if the complexity coefficient is positive and statistically significant (p < 0.05).**

This approach will allow us to test whether complexity predicts a game's average rating while controlling for its popularity.

###Analysis for Hypothesis 1:
We have found the **Pearson Correlation is 0.5518** and the **Spearman Correlation is 0.5337.** Since both of these values represent a **strong positive correlation**, therefore we can conclude that **as game complexity increases, the average rating also tends to increase.**

The coef value of complexity tells us **for every one point increase in complexity, the average rating increase by approximately 0.55 points**, and the p value (which is less than 0.001) suggests that this is **statistically significant.**

The variable log_num_ratings gave us a coefficient of -0.0156 and a p-value of 0.415. Because the p-value is greater than our significance level of 0.05, **we fail to reject the null hypothesis** for this variable. This indicates that, when **controlling for complexity, the number of ratings is not a statistically significant predictor of a board game's average rating.**

From the residual plot below, we can conclude that the linear line of best fit is a good model for complexity vs rating.

```python
pearson_corr, p_p = pearsonr(df['complexity_std'], df['avg_rating_std'])
spearman_corr, s_p = spearmanr(df['complexity_std'], df['avg_rating_std'])

print(f"Pearson Correlation (Complexity vs Avg Rating): {pearson_corr:.4f}")
print(f"Spearman Correlation (Complexity vs Avg Rating): {spearman_corr:.4f}")

X1 = df[['complexity_std', 'log_num_ratings_std']]
X1 = sm.add_constant(X1)
y = df['avg_rating_std']
model1 = sm.OLS(y, X1).fit()

print("--- Model 1 Summary (Linear Playtime) ---")
print(model1.summary())

plt.figure(figsize=(8, 10))

plt.subplot(2,1, 1)
sns.scatterplot(x = model1.fittedvalues, y= model1.resid, alpha=0.3, color='blue')
plt.axhline(0, color='red', linestyle='--')
plt.title('Model 1 Residuals')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')

plt.subplot(2, 1 ,2)
sns.regplot(x='complexity_std', y='avg_rating_std', data=df,
            scatter_kws={'alpha':0.1, 's':10}, line_kws={'color':'red'})
plt.title('Complexity vs Avg Rating')
plt.xlabel('Complexity')
plt.ylabel('Average Rating')

plt.tight_layout()
plt.savefig('model1_analysis.png')
plt.show()
```

##Hypothesis 2:
***Games with longer average playtimes receive lower average ratings, controlling for complexity.***

While challenge can be engaging, there is evidence that more time spent in games is not always linked to higher well-being or satisfaction. Vuorre et al. analyze large-scale gameplay data and find that total time spent playing video games has little to no direct impact on well-being compared to other factors like motivation and context (Vuorre et al.). This suggests that **raw time cost is not inherently positive.**

For tabletop games, longer playtime can be seen as a higher “investment” per session: even if players appreciate complexity, some may penalize games that take too long relative to the value they provide. Conditional on complexity, we therefore hypothesize that longer average playtime will be associated with slightly lower ratings.


**Null Hypothesis (H₀):**
 Average playtime has no effect on average BGG rating after controlling for complexity. In a linear regression, the coefficient for playtime = 0.

Because playtime_mid is likely skewed, we will also test a **log-transformed version (log_playtime_mid)** to see if it improves model fit and reduces heteroskedasticity, as discussed in class.

**We will reject H₀ only if the playtime coefficient is negative and statistically significant (p < 0.05) in at least one of the models after controlling for complexity.**

```python
X2 = df[['complexity_std', 'log_playtime_mid_std']]
X2 = sm.add_constant(X2)
y = df['avg_rating_std']
model2 = sm.OLS(y, X2).fit()

print("--- Model 2 Summary (Log-Transformed Playtime) ---")
print(model2.summary())

plt.figure(figsize= (8, 10))

plt.subplot(2,1, 1)
sns.scatterplot(x=model2.fittedvalues, y=model2.resid, alpha=0.3, color='green')
plt.axhline(0, color='red', linestyle='--')
plt.title('Model 2 Residuals')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')

plt.subplot(2,1, 2)
sns.regplot(x='log_playtime_mid', y='avg_rating', data = df,
            scatter_kws={'alpha':0.1, 's':10, 'color':'green'}, line_kws={'color':'orange'})
plt.title('Log(Playtime) vs Avg Rating')
plt.xlabel('Log(Average Playtime)')
plt.ylabel('Average Rating')

plt.tight_layout()
plt.savefig('model2_analysis.png')
plt.show()
```

#Data Limitations

While the BoardGameGeek dataset provides comprehensive coverage of thousands of board games with detailed statistics on ratings, complexity, and ownership, it still contains several limitations that may influence the interpretation of our analyses. These limitations mainly stem from how the data was collected (via web scraping) and the nature of the BGG community itself. Understanding these gaps is important for accurately framing the scope and generalizability of our results.

###Sampling Bias / User Representation
The dataset is sourced from the BoardGameGeek (BGG) community, which mainly consists of hobbyist and experienced players. Prior work and community descriptions suggest that these users tend to favor more complex, strategic, and Euro-style games over casual or family-oriented titles. As a result, any analyses involving ratings or popularity metrics are likely to reflect the preferences of an enthusiast subcommunity rather than the general market.

This has direct implications for our findings: our analysis for Hypothesis 1 shows a **strong positive association between complexity and average rating**, but this relationship may be partly driven by the self-selection of BGG users who already prefer deeper games. In other words, our estimates of the “benefit” of complexity are likely biased upward relative to what we would see in a more casual or mainstream player population. Consequently, highly rated games in this dataset may not be representative of broader public appeal.

###Data Reduction from Cleaning and Missingness
Several fields in the original dataset, such as minimum age, Amazon price, and designer information, contained missing or placeholder values. During preprocessing, we removed incomplete rows and excluded variables with excessive missingness to ensure data quality. While the final dataset is now complete for analysis, this cleaning process reduced the number of usable records.

Consequently, our results are based on a filtered subset of the original data that may **overrepresent well-documented or commercially available games**. This limits the generalizability of our conclusions to the broader set of games listed on BoardGameGeek.

###Playtime Measurement Approximation
To simplify analysis, we constructed playtime_mid as the average of the reported minimum and maximum playtimes for each game. However, some games have extremely wide playtime ranges (e.g., “30–180 minutes”), and **taking a simple midpoint may not accurately reflect typical session length or how most players experience the game**. For these titles, playtime_mid can be a noisy or misleading proxy for actual playtime, which may weaken our ability to detect a true relationship between playtime and rating.

This limitation is especially relevant for Hypothesis 2, where we test whether longer games receive lower ratings after controlling for complexity. If playtime_mid does not capture the “typical” experience, our estimates may be attenuated or unreliable. **A more robust approach would be to use a median-based measure of playtime** (if session-level or distributional data were available), or at least compare results using alternative definitions (e.g., the BGG “average playtime” field where it exists) to assess sensitivity. **In our current analysis, we note this measurement issue as a potential source of noise in the playtime–rating relationship.**

###Temporal Inconsistency
Ratings, ownership counts, and prices were scraped at different times, but the dataset does not include clear timestamps for when each metric was recorded. This inconsistency **limits any longitudinal or trend-based interpretations**. For example, we cannot tell whether changes in ratings reflect genuine shifts in popularity or simply the timing of data collection.

###Scraping and Data Quality Errors
Because the dataset was gathered through automated web scraping, some entries may contain parsing errors or outdated information due to changes in the BGG website structure. These inconsistencies can slightly **distort descriptive statistics** like mean ratings or ranking distributions and may introduce **noise** into predictive analyses.

###Duplicate or Merged Game Entries
BGG includes multiple editions and expansions of the same game under similar names, and it is likely that not all duplicates were merged during dataset preparation. We removed exact duplicate titles during cleaning, but it is possible that some near-duplicate entries remain (for example, “Catan,” “Catan: 5–6 Player Extension,” and “Catan (20th Anniversary Edition).” Duplicate listings can inflate popularity or ownership counts, making certain titles **appear more significant in correlation or ranking analyses than they actually are**. After conducting research on how to tackle this issue, we concluded that any further data deduplocation is outside the scope of this research endeavor. Instead, we leave this as future work that can extend from our project.  


#Conclusion

Our project set out to understand **which factors best predict a game’s average user rating on BoardGameGeek (BGG).** Using a cleaned subset of modern titles with complete rating, complexity, and playtime information, we focused on two preregistered hypotheses: **that higher complexity is associated with higher ratings (Hypothesis 1),** and that **longer playtime is associated with lower ratings once complexity is held constant (Hypothesis 2)**.

For Hypothesis 1, our correlation and regression results consistently show that **complexity is a strong, positive predictor of average rating.** Both Pearson and Spearman correlations between complexity and rating are moderate and positive, and in our multiple regression of average rating on complexity and log-transformed number of ratings, the coefficient on complexity is large and highly statistically significant, while the popularity term (log number of ratings) is small and not significant. Substantively, this means that, **for the BGG hobbyist audience, moving up one point on the 1–5 complexity scale is associated with a notable increase in average rating, even after controlling for how many people rated the game.** This supports our hypothesis that **BGG users reward games that are more strategically demanding** and confirms that **“heavier” games tend to be rated more favorably than lighter ones.**

For Hypothesis 2, our analyses **do not** provide evidence that longer games are penalized once complexity is taken into account. In models that include both complexity and playtime (in either minutes or log-transformed playtime), the estimated playtime coefficient is extremely small and not robustly statistically significant, while complexity remains positive and highly significant. The corresponding scatterplots are also visually flat with respect to playtime once complexity is considered. Together, these results lead us to **reject** Hypothesis 2: within this dataset, **we do not find a meaningful linear relationship between playtime and rating once complexity is controlled for.**

Putting these pieces together, the **clearest predictor of higher BGG ratings in our models is complexity, not session length.** Among the predictors we examined, complexity shows the strongest and most stable association with ratings, while playtime and popularity measures play at most a secondary role. This pattern fits the intuition that BGG’s self-selected community of hobbyists values cognitive depth and strategic richness more than the raw time cost of a game. At the same time, because our data come from an **enthusiast** platform rather than a **representative consumer sample,** **these conclusions are most appropriate for designers and publishers targeting the BGG-style audience rather than the general market.**

In a broader sense, our findings speak to **how niche online communities can shape what “good” looks like in the world of consumer products.** For game designers and publishers, our results suggest that, at least within the hobbyist market, investing in depth and meaningful decisions may be more important for perceived quality than keeping playtime short. For marketers and platforms, our study highlights how ratings drawn from self-selected enthusiasts can systematically differ from mainstream preferences, which has implications for how recommendation systems are designed and how companies interpret user-generated ratings when forecasting demand. More generally, our project illustrates the **importance of thinking critically about who is in a dataset** and how that shapes the **“truths” we infer about the world.**

Overall, our analyses answer the opening research question by showing that, **for modern board games on BoardGameGeek, increased complexity reliably predicts higher user ratings, while longer playtime does not systematically help or hurt once complexity is taken into account**. Future work could extend this by testing **non-linear playtime effects, incorporating genre or theme, or combining BGG data with more mainstream sales or review sources** to see whether the “complex games do better” result generalizes **beyond this enthusiast community.** Taken as a whole, our findings suggest that, at least within BGG and similar hobbyist spaces, making games deeper rather than simply longer is what most consistently pays off in terms of user ratings, and that understanding who is doing the rating is crucial when using data to make decisions in the real world.
